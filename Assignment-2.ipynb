```python
#Solution 1 – Student Report Card
students = {"Rahul": 91,"Ayesha": 84,"Kiran": 67,"Ali": 45}

pass_count = 0
fail_count = 0

for name, marks in students.items():
    print("Name:", name)
    print("Marks:", marks)

    if marks >= 40:
        print("Result: Pass")
        pass_count += 1
    else:
        print("Result: Fail")
        fail_count += 1

    print("----------------")

print("Total Passed:", pass_count)
print("Total Failed:", fail_count)
```

```python
#Solution 2 – Grocery Store Inventory
inventory = {"Rice": 20,"Sugar": 15,"Milk": 8,"Eggs": 30}

item = "Milk"

if item in inventory:
    inventory[item] -= 1
    print(item, "Purchased Successfully")
else:
    print("Item Not Available")

print("Updated Inventory:")
print(inventory)
```

```python
#Solution 3 – Login System
users = {"admin": "python123","student": "welcome","guest": "guest123"}

attempt = 1

while attempt <= 3:
    username = input("Enter Username: ")
    password = input("Enter Password: ")

    if username in users and users[username] == password:
        print("Login Successful")
        break
    else:
        print("Invalid Username or Password")
        attempt += 1

if attempt > 3:
    print("Account Locked")
```

```python
#Solution 6 – Daily Expense Tracker
expenses = {}

while True:
    name = input("Enter Expense Name (or stop): ")

    if name.lower() == "stop":
        break

    amount = float(input("Enter Amount: "))
    expenses[name] = amount

print("\nExpense Details")

for item, price in expenses.items():
    print(item, ":", price)

print("Total Expenses:", len(expenses))
```

```python
#Solution 10 – Safe Calculator
try:
    num1 = float(input("Enter First Number: "))
    num2 = float(input("Enter Second Number: "))

    result = num1 / num2

    print("Answer:", result)

except ValueError:
    print("Invalid Number Entered")

except ZeroDivisionError:
    print("Cannot Divide By Zero")
```

```python
#Solution 11 – High Score Board
scores = {"Rahul": 120,"Ali": 250,"John": 180}

highest_player = ""
highest_score = 0

for player, score in scores.items():
    if score > highest_score:
        highest_score = score
        highest_player = player

print("Highest Player:", highest_player)
print("Highest Score:", highest_score)
```

```python
#Solution 13 – Movie Ticket Booking
seats = {"A1": "Available","A2": "Booked","A3": "Available","A4": "Available"}

seat = input("Enter Seat Number: ").upper()

if seat in seats:
    if seats[seat] == "Available":
        seats[seat] = "Booked"
        print("Seat Booked Successfully")
    else:
        print("Seat Already Booked")
else:
    print("Invalid Seat Number")

print(seats)
```

```python
#Solution 15 – Shopping Bill Generator
items = {"Milk": 50,"Bread": 40,"Eggs": 70,"Rice": 500}

bill = 0

while True:
    item = input("Enter Item Name (done to stop): ").title()

    if item.lower() == "done":
        break

    if item in items:
        bill += items[item]
        print(item, "Added")
    else:
        print("Item Not Available")

print("Final Bill:", bill)
```

```python
#Solution 20 – Space Mission Log
systems = {"Engine": "OK","Fuel": "OK","Navigation": "FAIL","Communication": "OK"}

file = open("mission_log.txt", "w")

for system, status in systems.items():
    if status == "FAIL":
        print(system, "Failed")
        file.write(system + " Failed\n")
        print("Mission Aborted")
        break
else:
    file.write("Launch Successful")

file.close()
```
