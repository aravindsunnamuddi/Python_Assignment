```python
# Problem 1 – Secret Diary

memory = input("Enter today's memory: ")

file = open("diary.txt", "a")
file.write(memory + "\n")
file.close()

file = open("diary.txt", "r")
print("\nAll Saved Memories:")
print(file.read())
file.close()
```

```python
# Problem 3 – Shopping History

file = open("orders.txt", "r")

orders = file.readlines()

print("Items Purchased:")

for item in orders:
    print(item.strip())

print("Total Orders:", len(orders))

file.close()
```

```python
# Problem 4 – Password Inspector

password = input("Enter Password: ")

upper = 0
lower = 0
digit = 0

for ch in password:
    if ch.isupper():
        upper += 1
    elif ch.islower():
        lower += 1
    elif ch.isdigit():
        digit += 1

print("Uppercase Letters:", upper)
print("Lowercase Letters:", lower)
print("Digits:", digit)

if len(password) >= 8 and upper > 0 and lower > 0 and digit > 0:
    print("Strong Password")
else:
    print("Weak Password")
```

```python
# Problem 6 – Flight Ticket

city = input("Enter City Name: ")

city = city.strip().title()

print("Cleaned City Name:", city)
```

```python
# Problem 7 – ATM Withdrawal

def withdraw(balance, amount):
    if amount <= balance:
        balance -= amount
        return balance
    else:
        return "Insufficient Balance"

balance = 10000

amount = int(input("Enter Withdrawal Amount: "))

result = withdraw(balance, amount)

print("Result:", result)
```

```python
# Problem 8 – Grade Calculator

def calculate_grade(mark):
    if mark >= 90:
        return "A"
    elif mark >= 75:
        return "B"
    elif mark >= 60:
        return "C"
    else:
        return "F"

mark = int(input("Enter Marks: "))

grade = calculate_grade(mark)

print("Grade:", grade)
```

```python
# Problem 9 – Bill Calculator

def calculate_bill(price, quantity):
    return price * quantity

price = float(input("Enter Price: "))
quantity = int(input("Enter Quantity: "))

bill = calculate_bill(price, quantity)

print("Total Bill:", bill)
```

```python
# Problem 10 – Discount Calculator

discount = lambda price: price - (price * 10 / 100)

price = float(input("Enter Product Price: "))

print("Price After Discount:", discount(price))
```

```python
# Problem 11 – Rectangle Area

area = lambda length, width: length * width

length = int(input("Enter Length: "))
width = int(input("Enter Width: "))

print("Area of Rectangle:", area(length, width))
```

```python
# Problem 12 – Speed Checker

speed_check = lambda speed: "Overspeed" if speed > 80 else "Normal Speed"

speed = int(input("Enter Speed: "))

print(speed_check(speed))
```
